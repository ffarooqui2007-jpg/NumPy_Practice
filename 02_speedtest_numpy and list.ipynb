```python
# using list method

import numpy as np
import time

size=1_000_000
l1=list(range(size))
l2=list(range(size))
start=time.time()
add=[x+y for x,y in zip(l1,l2)]
end=time.time()

print(add[:10])
print(end - start)
```

```python
# using numpy library

import numpy as np  # u can also use import python as ff or whatever u want.
import time 

size=1_000_000
l1=np.array(range(size))
l2=np.array(range(size))
start=time.time()
add=l1+l2 
end=time.time()

print(add[:10])
print(end-start)
```
