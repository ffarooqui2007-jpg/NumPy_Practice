```python
# numpy suppoert vectorized operation
import time
lst=[1,2,3,4,5,7,8]
start=time.time()
new_lst=[x**2 for x in lst]
end=time.time()
print(new_lst)
print(end-start)
```

```python
import matplotlib.pyplot as plt

# Line chart
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 210, 195]

# Bar chart
products = ["Apples", "Bananas", "Oranges", "Grapes"]
quantities = [35, 50, 28, 42]

# Scatter plot
study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
test_scores = [52, 58, 65, 68, 75, 78, 86, 91]

# Histogram
daily_temperatures = [18, 20, 22, 21, 19, 24, 25, 23, 22, 20,
                      19, 21, 26, 27, 24, 22, 23, 20, 18, 25]

fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].plot(months, sales, marker="o")
axes[0, 0].set_title("Monthly Sales")

axes[0, 1].bar(products, quantities)
axes[0, 1].set_title("Product Quantities")

axes[1, 0].scatter(study_hours, test_scores)
axes[1, 0].set_title("Study Hours vs Test Scores")
axes[1, 0].set_xlabel("Study hours")
axes[1, 0].set_ylabel("Test score")

axes[1, 1].hist(daily_temperatures, bins=6, edgecolor="black")
axes[1, 1].set_title("Daily Temperatures")

plt.tight_layout()
plt.show()
```
